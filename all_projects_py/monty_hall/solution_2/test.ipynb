```python
import random


def monty_hall_game(switch_doors):
    doors = ['car', 'goat', 'goat']

    random.shuffle(doors)

    initial_choice = random.choice(range(3))

    if switch_doors:
        doors_revaled = [i for i in range(3) if i != initial_choice and doors[i] != 'car']

        door_revaled = random.choice(doors_revaled)
        
        final_choice = [i for i in range(3) if i != initial_choice and i != door_revaled][0]  # noqa: RUF015
    else:
        final_choice = initial_choice

    return doors[final_choice] == 'car'
```

```python
def simulate_game(num_games):
    num_wins_without_switch = sum([monty_hall_game(False) for _ in range(num_games)])
    num_wins_with_switch = sum([monty_hall_game(True) for _ in range(num_games)])
    return num_wins_without_switch / num_games, num_wins_with_switch / num_games
```

```python
simulate_game(10000)
```
