- put back of the doors randomly
- choose one of them
- put the one of two back
- change or stay my choose
- repeat 1000 times

```python
import random


def put_randomly():
    doors = {
        1: None,
        2: None,
        3: None
    }

    values = ["goat", "car", "goat"]

    random.shuffle(values)

    for door, value in zip(doors, values):
        doors[door] = value

    return doors


def choose_door():
    
    doors = put_randomly()
    doors2 = doors.copy()
    choice = int(input('choose you door between 1, 2, 3 : '))
    if choice == 1:
        del doors[1]
    if choice == 2:
        del doors[2]
    if choice == 3:
        del doors[3]

    value = "goat"

    for key, val in doors.items():
        if val == value:
            print(f'this door ({key}) has a goat')
            break
    return key, doors2, choice

def change_door():
    key, doors2, choice = choose_door()
    del doors2[key]
    decide = input('do you want to change your choice ? y/n : ')
    if decide == "y":
        for key, value in doors2.items():
            if key != choice:
                print(f"you choose the door {key} and it was a {value}")

    else:
        print(f"you choose the door {choice} and it was a {doors2[choice]}")
```

```python
change_door()
# print(doors)
```
