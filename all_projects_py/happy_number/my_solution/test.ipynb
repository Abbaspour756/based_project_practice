```python
def seprating(number):
    a = []
    while number > 0:
        digit = number % 10
        number //= 10
        a.append(digit)
    return a
```

```python
def finder(number, count=0):

    if count >= 8:
       return "Stopped after 8 attempts"

    while number != 1:
        num = seprating(number)
        jam = 0
        for i in num:
            jam += i ** 2
        print(jam)
        return finder(jam, count + 1)
    return "This is a happy number"
```
