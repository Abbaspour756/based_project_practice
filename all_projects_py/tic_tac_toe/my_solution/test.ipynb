```python
class TicTacToe:
    def __init__(self):
        self.board = [[' ',' ', ' '], [' ', ' ', ' '], [' ', ' ', ' ']]
    
    def show_board(self):
        for i in self.board:
            print(i)
    
    def circle(self):
        target = int(input('player O, choose a cell: '))
        row = (target - 1) // 3
        col = (target - 1) % 3
        if self.board[row][col] == ' ':
            self.board[row][col] = 'O'

        else:
            print('cell is already occupied, try again')
            self.circle()

    def cross(self):
        target = int(input('player X, choose a cell: '))
        row = (target - 1) // 3
        col = (target - 1) % 3

        if self.board[row][col] == ' ':
            self.board[row][col] = 'X'
        else:
            print('cell is already occupied, try again')
            self.cross()
def start():
    game = TicTacToe()
    game.show_board()
    win = True
    while win:
        if game.board[0][0] == game.board[0][1] == game.board[0][2] == ('O' or 'X'):
            win = False
            print('player', game.board[0][0], 'wins')
        if game.board[1][0] == game.board[1][1] == game.board[1][2] == ('O' or 'X'):
            win = False
            print('player', game.board[1][0], 'wins')
        if game.board[2][0] == game.board[2][1] == game.board[2][2] == ('O' or 'X'):
            win = False
            print('player', game.board[2][0], 'wins')
        if game.board[0][0] == game.board[1][0] == game.board[2][0] == ('O' or 'X'):
            win = False
            print('player', game.board[0][0], 'wins')
        if game.board[0][1] == game.board[1][1] == game.board[2][1] == ('O' or 'X'):
            win = False
            print('player', game.board[0][1], 'wins')
        game.circle()
        game.cross()
        game.show_board()
    
```

```python
start()
```
